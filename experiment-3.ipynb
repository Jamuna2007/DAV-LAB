```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("student.csv.txt")

sns.histplot(df["Marks"])
plt.title("Marks Distrubution")
plt.show()

sns.boxplot(x="Department" , y="Marks", data=df);
plt.title("Marks by Branch")
plt.show()

sns.scatterplot(x="Attendance", y="Marks" , hue="Department",  data=df)
plt.title("Attendance vs Marks")
plt.show()

sns.pairplot(df[["Attendance", "Marks", "Department", "Gender", "Age"]],hue="Gender")
plt.show()

sns.heatmap(df.corr(numeric_only=True),annot=True)
plt.title("Correlation Heatmap ")
plt.show()
```
