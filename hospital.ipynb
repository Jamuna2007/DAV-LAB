```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("healthcare.csv.txt")


sns.histplot(df["Blood_Sugar"])
plt.title("Blood Sugar Distribution")
plt.show()


sns.boxplot(x="Disease", y="BMI", data=df)
plt.title("BMI by Disease")
plt.show()


sns.scatterplot(x="Cholesterol", y="Blood_Sugar", hue="Disease", data=df )
plt.title("Cholesterol vs Blood Sugar")
plt.show()


sns.pairplot( df[["Age", "Cholesterol", "Blood_Sugar", "BMI", "Disease"]], hue="Disease")
plt.show()

sns.heatmap( df.corr(numeric_only=True),annot=True,cmap="coolwarm")
plt.title("Healthcare Correlation Heatmap")
plt.show()
```
